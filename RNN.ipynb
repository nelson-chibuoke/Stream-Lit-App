```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers

#Make the results same thing every time
np.random.seed(42)
keras.utils.set_random_seed(42)
```

```python
memory = 0.0

for step in range(1, 4):
    new_input = 1.0
    memory = np.tanh(0.5 * new_input + 0.8 * memory)
    print(f"Step {step}: input = {new_input} -> memory = {memory:.3f}")
```

```python
time = np.arange(0, 400)
wave = np.sin(time * 0.1)

plt.figure(figsize=(10, 3))
plt.plot(time[:150], wave[:150])
plt.title("Our wave (first 150 points)")
plt.xlabel("Time step")
plt.ylabel("Value")
plt.show()
```

```python
window = 20

x = []
y = []
for i in range(len(wave) - window):
    x.append(wave[i : i + window])
    y.append(wave[i + window])

x = np.array(x)
y = np.array(y)

x = x.reshape(-1, window, 1)

print("x shape:", x.shape)
print("y.shape:", y.shape)
```
